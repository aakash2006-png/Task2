A* algorithm

```python
import heapq

# Graph: node -> {neighbor: cost}
graph = {
    'S': {'A': 3, 'C': 2},
    'A': {'B': 4},
    'B': {'D': 5},
    'C': {'D': 2},
    'D': {}
}

# Heuristic values: h(n)
heuristic = {
    'S': 4,
    'A': 7,
    'B': 5,
    'C': 2,
    'D': 0
}

start = 'S'
goal = 'D'

# Priority queue: (f_cost, g_cost, node)
open_list = []
heapq.heappush(open_list, (heuristic[start], 0, start))

# Cost from start to each node
g_cost = {start: 0}

# Parent of each node for path reconstruction
parent = {start: None}

# Nodes that have been explored
explored = []

while open_list:

    # Get node with smallest f = g + h
    f, g, current = heapq.heappop(open_list)

    # Ignore outdated entries
    if g != g_cost.get(current, float('inf')):
        continue

    explored.append(current)

    print(
        f"Node: {current} | "
        f"g = {g} | "
        f"h = {heuristic[current]} | "
        f"f = {f}"
    )

    # Goal reached
    if current == goal:
        break

    # Explore neighbors
    for neighbor, cost in graph[current].items():

        new_g = g + cost

        # Found a cheaper path
        if new_g < g_cost.get(neighbor, float('inf')):

            g_cost[neighbor] = new_g
            parent[neighbor] = current

            new_f = new_g + heuristic[neighbor]

            heapq.heappush(
                open_list,
                (new_f, new_g, neighbor)
            )

# Check whether goal was reached
if goal not in g_cost:
    print("\nNo path found.")

else:
    # Reconstruct path
    path = []
    current = goal

    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()

    print("\nNodes explored:", " -> ".join(explored))
    print("Final path:", " -> ".join(path))
    print("Total cost:", g_cost[goal])
```


hill climbing

```python
def f(x):
    return (x - 5) ** 2 + 10

x = 0
step = 1

print("Hill Climbing Search")
print("--------------------")

while True:
    current_value = f(x)

    left = x - step
    right = x + step

    left_value = f(left)
    right_value = f(right)

    print("x =", x, ", f(x) =", current_value)

    if left_value < current_value and left_value <= right_value:
        x = left

    elif right_value < current_value:
        x = right

    else:
        break

print("\nFinal solution:")
print("x =", x)
print("Minimum f(x) =", f(x))
```
